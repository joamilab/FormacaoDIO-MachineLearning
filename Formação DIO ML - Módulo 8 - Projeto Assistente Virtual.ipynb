## Módulo 1

Transformação de texto em fala

```python
from gtts import gTTS
from IPython.display import Audio
```

```python
def speak_text(text, language):
  '''
    Converte texto em áudio.

    Params:
      text (str): Texto a ser convertido.
      language (str): Idioma do texto.
    Returns:
      Audio: Áudio do texto.
  '''

  gtts_object = gTTS(text = text,
                     lang = language,
                     slow = False)

  gtts_object.save("audio-ajuda.wav")

  return Audio("audio-ajuda.wav", autoplay=True)
```

```python
text = "Como posso te ajudar?"
language = "pt"
  
audio = speak_text(text, language)
audio
```

## Módulo 2
* Transformação de fala em texto
* Acionar função com comando de voz

```python
import speech_recognition as sr
from gtts import gTTS
import os
from datetime import datetime
import playsound
import wikipedia
import pyaudio
import webbrowser
```

```python
#speak converted audio to text
def speak(text, language):
    tts = gTTS(text=text, lang='en')
    filename = "voice.mp3"
    try:
        os.remove(filename)
    except OSError:
        pass
    tts.save(filename)
    playsound.playsound(filename)

#get mic audio
def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        r.pause_threshold = 1
        # wait for a second to let the recognizer adjust the
        # energy threshold based on the surrounding noise level
        r.adjust_for_ambient_noise(source, duration=1)
        audio = r.listen(source)
        said = ""
        try:
            said = r.recognize_google(audio, language="pt-BR")
            print(said)
        except sr.UnknownValueError:
            speak("Desculpe, eu não entendi.", "pt")
        except sr.RequestError:
            speak("Desculpe, o serviço não está disponível.", "pt")
    return said.lower()
```

```python
#function to respond to commands
def respond(text, language):

    print("Texto transcrito do áudio " + text)

    if 'youtube' in text:
        speak("O que você quer buscar?", language)
        keyword = get_audio()
        if keyword!= '':
            url = f"https://www.youtube.com/results?search_query={keyword}"
            webbrowser.get().open(url)
            speak(f"Aqui está o que encontramos para a busca {keyword} no youtube", language)

    elif 'pesquisar' in text:
        speak("O que você quer pesquisar?", language)
        query = get_audio()
        if query !='':
            result = wikipedia.summary(query, sentences=3)
            speak("De acordo com a Wikipedia", language)
            print(result)
            speak(result, language)

    elif 'horas' in text:
        strTime = datetime.today().strftime("%H:%M %p")
        print(strTime)
        speak(strTime, language)

    elif 'sair' in text:
        speak("Tchau, até a próxima!", language)
        return
```

```python
while True:
    print("Estou ouvindo...")
    text = get_audio()
    respond(text, "pt")
```
