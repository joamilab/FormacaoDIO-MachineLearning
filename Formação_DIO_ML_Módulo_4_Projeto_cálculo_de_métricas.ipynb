# Cálculo de métricas sem o uso de bibliotecas

### Carregar bibliotecas para buscar dataset e treinar o modelo

```python
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
```

### Carregar dataset de classificação de casos de câncer de mama

```python
data = load_breast_cancer()
data.data.shape
```

### Calcular quantidade de amostras que serão usadas no treino e no teste do modelo

```python
len_train = int(data.data.shape[0]*0.7)
len_test = data.data.shape[0] - len_train

print(len_train, len_test)
```

### Treino do modelo com algoritmo de Árvore de Decisão

```python
model = DecisionTreeClassifier().fit(data.data[:len_train], data.target[:len_train])
```

### Predição das amostras de teste usando o modelo treinado

```python
y_true = data.target[len_train:]
print(y_true[:10])

y_pred = model.predict(data.data[len_train:])
print(y_pred[:10])
```

### Contagem de acertos e erros
* VP: Verdadeiro Positivo
* VN: Verdadeiro Negativo
* FP: Falso Positivo
* FN: Falso Negativo

```python
VP, VN, FP, FN = 0, 0, 0, 0

for i in range(len(y_true)):
  if y_true[i] == 1 and  y_pred[i] == 1:
    VP += 1
  elif y_true[i] == 1 and  y_pred[i] == 0:
    FN += 1
  elif y_true[i] == 0 and  y_pred[i] == 1:
    FP += 1
  elif y_true[i] == 0 and  y_pred[i] == 0:
    VN += 1
```

### Cálculo das métricas

```python
acuracia = (VP + VN) / (VP + VN + FP + FN)
precisao = VP / (VP + FP)
sensibilidade = VP / (VP + FN)
especificidade = VN / (VN + FP)
f1_score = 2 * (precisao * sensibilidade) / (precisao + sensibilidade)
```

```python
print('--- MATRIZ DE CONFUSÃO ---\n')
print(f'[{VP}, {FN}]')
print(f'[{FP}, {VN}]')

print('\n-------- MÉTRICAS --------\n')
print('Acurácia: {:.5f}'.format(acuracia))
print('Precisão: {:.5f}'.format(precisao))
print('Sensibilidade: {:.5f}'.format(sensibilidade))
print('Especificidade: {:.5f}'.format(especificidade))
print('F1 Score: {:.5f}'.format(f1_score))
```
